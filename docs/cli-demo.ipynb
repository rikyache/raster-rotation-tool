# RasterRotation CLI demo

Замените `C:\\Images\\photo.jpg` на путь к реальному изображению перед запуском.

```powershell
dotnet run --project ..\src\RasterRotation.Cli -- "C:\Images\photo.jpg" --info
```

```powershell
dotnet run --project ..\src\RasterRotation.Cli -- "C:\Images\photo.jpg" --rotate 90 --output "C:\Images\photo_rotated.jpg"
```

```powershell
dotnet run --project ..\src\RasterRotation.Cli -- "C:\Images\photo.jpg" --crop 20,20,640,480 --flip h --output "C:\Images\photo_crop.jpg"
```
